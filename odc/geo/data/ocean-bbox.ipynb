```python
import json

import antimeridian

with open("ocean.geojson") as f:
    ocean = json.load(f)

print(f"The ocean feature collection has {len(ocean['features'])} features")
print(
    f"The first bounding box is: {antimeridian.bbox(ocean['features'][0]['geometry'])}"
)
print(
    f"The second bounding box is: {antimeridian.bbox(ocean['features'][1]['geometry'])}"
)

from odc.geo.geom import multigeom, Geometry

gg = multigeom([Geometry(f["geometry"], "epsg:4326") for f in ocean["features"]])

print(f"The bbox of the ocean feature multipolygon is: {antimeridian.bbox(gg)}")
```
